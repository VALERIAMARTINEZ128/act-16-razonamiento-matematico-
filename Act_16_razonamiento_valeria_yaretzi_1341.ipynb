# Problema 5: Dimensiones del terreno
# perimetro = 50
# diferencia_largo = 5
perimetro = int(input("Ingrese el perímetro: "))
diferencia_largo = int(input("Ingrese la diferencia del largo: "))

# Ecuación simplificada: 4x + (2 * diferencia) = perimetro
# x = (perimetro - 2 * diferencia) / 4
ancho = (perimetro - (2 * diferencia_largo)) / 4
largo = ancho + diferencia_largo

print(f"El ancho mide: {ancho} metros")
print(f"El largo mide: {largo} metros")

```python
def calcular_dimensiones():
    print("--- Calculadora de Dimensiones de Terreno ---")

    try:
        # Solicitamos los datos al usuario
        perimetro = float(input("Ingrese el perímetro total (m): "))
        diferencia = float(input("¿Cuánto más mide el largo respecto al ancho? (m): "))

        # Aplicamos tu lógica:
        # 4x + 2d = P  =>  x = (P - 2d) / 4
        ancho = (perimetro - (2 * diferencia)) / 4
        largo = ancho + diferencia

        # Verificación de lógica física (el ancho no puede ser negativo)
        if ancho <= 0:
            print("\nError: Las medidas proporcionadas no forman un rectángulo real.")
        else:
            print("-" * 30)
            print(f"Resultados:")
            print(f"  Ancho: {ancho:>10.2f} metros")
            print(f"  Largo: {largo:>10.2f} metros")
            print(f"  Área:  {ancho * largo:>10.2f} m²")
            print("-" * 30)

    except ValueError:
        print("Error: Por favor, ingrese solo valores numéricos.")

# Ejecutar la función
if __name__ == "__main__":
    calcular_dimensiones()
```
